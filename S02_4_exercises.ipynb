### Exercise 1.1: Counting Words

Given the sentence from the slides: *"They picnicked by the pool, then lay back on the grass and looked at the stars."*

Write Python code to:
1. Count words excluding punctuation
2. Count words including punctuation as separate tokens
3. Count unique word types (case-insensitive)

````python
import re
# YOUR CODE HERE
sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."

# 1 
words_only = re.findall(r'\w+', sentence)
print(f"Palabras sin puntuación: {len(words_only)}")

# 2
all_tokens = re.findall(r'\w+|[^\w\s]', sentence)
print(f"Tokens con puntuación: {len(all_tokens)}")

# 3
unique_types = set(word.lower() for word in words_only)
print(f"Tipos únicos: {len(unique_types)}")
````

### Exercise 1.2: Handling Disfluencies

The slides show the utterance: *"I do uh main- mainly business data processing"*

Write code to:
1. Count all tokens including disfluencies
2. Remove filled pauses (uh, um) and fragments (words ending with -)
3. Count "clean" words

````python
# YOUR CODE HERE
utterance = "I do uh main- mainly business data processing"
# 1
all_ut_tokens = utterance.split()
print(f"Total tokens: {len(all_ut_tokens)}")

# 2 y 3
clean_words = [w for w in all_ut_tokens if w.lower() not in ['uh', 'um'] and not w.endswith('-')]
print(f"Palabras limpias: {clean_words}")
print(f"Conteo limpio: {len(clean_words)}")
````

### Exercise 2.1: Basic Pattern Matching

Using the regex patterns from the slides, write code to:
1. Find all words starting with capital letters
2. Find all digits in a text
3. Find words that are NOT capitalized

````python
# YOUR CODE HERE
text = "Chapter 1: Down the Rabbit Hole. Alice was 7 years old in 1865."
capitals = re.findall(r'[A-Z]\w*', text) #  [A-Z]\w*, finds words that start with a capital letter, like names or the first word of a sentence.
print(capitals)
digits = re.findall(r'\d+', text)  # \d+, finds sequences of one or more digits, like numbers or years.
print(digits)
not_capitals = re.findall(r'\b[a-z]\w*', text) # [a-z]\w*, finds words that start with a lowercase letter and the \b marks a word boundary to ensure we only get whole words that are not capitalized.
print(not_capitals)
````

### Exercise 2.2: Kleene Star and Plus

The slides explain Kleene operators (* and +).

Write patterns to match:
1. "baa" followed by zero or more 'a's (baa, baaa, baaaa...)
2. One or more digits
3. Any sequence of characters (wildcard)

````python
# YOUR CODE HERE
test_strings = ["ba", "baa", "baaa", "baaaa", "baaaaa"]
# baa* matches 'ba' followed by ONE 'a', then zero or more 'a's
# The prompt asks for "baa" followed by zero or more
baa_pattern = r'baaa*' 
print("Matches for baa_pattern:")
for s in test_strings:
    if re.fullmatch(baa_pattern, s):  # fullmatch make sure the whole string matches
        print(s)
````

### Exercise 2.3: Anchors and Word Boundaries

Write patterns using anchors (^, $, \b) to:
1. Match lines starting with a capital letter
2. Match lines ending with a period
3. Find the word "the" as a complete word (not in "other" or "there")

````python
# YOUR CODE HERE
lines = [
    "The quick brown fox",
    "jumped over the lazy dog.",
    "other animals watched.",
    "There was nothing else."
]
# The ^ symbol is for the start of the line, so ^[A-Z] finds lines that begin with a capital letter. 
# The $ symbol is for the end of the line, so \.$ matches lines that finish with a period. 
# To find the word "the" , we use \bthe\b, the \b marks a word boundary, to make sure we don’t make the mistake to match “the” inside another word like “other” or “there.”
pattern_start_cap = r'^[A-Z]'
matches_start_cap = [line for line in lines if re.search(pattern_start_cap, line)]
print("Lines starting with a capital letter:", matches_start_cap)
pattern_end_period = r'\.$'
matches_end_period = [line for line in lines if re.search(pattern_end_period, line)]
print("Lines ending with a period:", matches_end_period)
pattern_word_the = r'\bthe\b'
matches_word_the = [line for line in lines if re.search(pattern_word_the, line, re.IGNORECASE)]
print('Lines containing the word "the":', matches_word_the)
````

### Exercise 2.4: Substitutions and Capture Groups

The slides show date format conversion. Write code to:
1. Convert US dates (mm/dd/yyyy) to EU format (dd-mm-yyyy)
2. Anonymize email addresses by replacing with [EMAIL]
3. Convert "I'm" contractions to "I am"

````python
# YOUR CODE HERE
text_dates = "Meeting on 10/15/2011 and 03/22/2024"
text_emails = "Contact john@example.com or jane@test.org"
text_contractions = "I'm happy and I'm excited"
````

### Solution 2.4

````python


# Use re.sub with the pattern (\d{2})/(\d{2})/(\d{4}) to capture the month, day, and year separately, then rearrange them into European format 15-10-2011.
text_dates = "The date is 10/15/2011"
converted = re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\2-\1-\3', text_dates)
print(f"1. US to EU date: '{text_dates}' -> '{converted}'")

# Use re.sub in a email with the pattern \b[\w.-]+@[\w.-]+\.\w+\b and replace them with [EMAIL]
text_emails = "Contact john@example.com or jane@test.org"
anon = re.sub(r'\b[\w.-]+@[\w.-]+\.\w+\b', '[EMAIL]', text_emails)
print(f"\n2. Anonymized: '{anon}'")

# Use re.sub to find contractions like I'm and replace them with their expanded forms like I am.
text_contractions = "I'm happy and I'm excited"
expanded = re.sub(r"I'm", "I am", text_contractions)
print(f"\n3. Expanded: '{expanded}'")
````

### Exercise 2.5: ELIZA-style Pattern Matching

The slides show ELIZA, an early chatbot using regex.
Implement simple ELIZA rules for:
1. "I'm [emotion]" -> "WHY DO YOU THINK YOU ARE [emotion]"
2. "I need [something]" -> "WHAT WOULD IT MEAN IF YOU GOT [something]"
3. Any input with "always" -> "CAN YOU THINK OF A SPECIFIC EXAMPLE?"

````python
# YOUR CODE HERE
def simple_eliza(user_input):
    #  "I'm [emotion]"
    match = re.match(r"I'm (.+)", user_input, re.IGNORECASE)
    if match:
        emotion = match.group(1)
        return f"WHY DO YOU THINK YOU ARE {emotion.upper()}?"

    # "I need [something]"
    match = re.match(r"I need (.+)", user_input, re.IGNORECASE)
    if match:
        something = match.group(1)
        return f"WHAT WOULD IT MEAN IF YOU GOT {something.upper()}?"

    # any input containing "always"
    if re.search(r"\balways\b", user_input, re.IGNORECASE):
        return "CAN YOU THINK OF A SPECIFIC EXAMPLE?"

    # Default response
    return "TELL ME MORE."
# Use re.match with the pattern "I'm (.+)" to see if the input starts with “I’m” , if it does, the text after it is captured and reused in a response about user’s feelings. 
# Do the same in the next part with "I need (.+)"  to catch what the user says they need and turn it into a reflective question. 
# Use re.search instead of re.match to check whether the word “always” appears anywhere in the sentence, and responds by asking for a specific example.
# Test
test_inputs = [
    "I'm depressed",
    "I need a car",
    "They always ignore me"
]

for t in test_inputs:
    print(simple_eliza(t))
````

### Exercise 3.1: Unicode Code Points

The slides explain that Unicode assigns code points to characters.
Write code to:
1. Get the code point of 'a' (should be U+0061)
2. Get the character for code point U+00F1 (ñ)
3. Show the UTF-8 byte encoding of "hello"

````python
# YOUR CODE HERE
# 1. With ord('a') we get the Unicode code point of the character 'a', which is a number that represents that character and we print it in a hexadecimal form.
code_point_a = ord('a')
print(f"'a' -> U+{code_point_a:04X}")

# 2. With chr(0x00F1) we do the opposite ,take a Unicode code point (U+00F1) and converts it back into the character that represents.
char_enye = chr(0x00F1)
print(f"U+00F1 -> {char_enye}")

# 3. With "hello".encode("utf-8") is an example of how a string is kept in UTF-8 by converting it into bytes to transmit or store it.
utf8_bytes = "hello".encode("utf-8")
print(f'"hello" in UTF-8 -> {utf8_bytes}')
````

### Exercise 3.2: Multi-byte UTF-8 Characters

The slides show that different scripts need different byte lengths.
Examine the byte lengths for:
1. ASCII character: 'A'
2. Spanish: 'ñ' (U+00F1)
3. Chinese: '姚' (a character from the slides)
4. Emoji: '😀'

````python
# YOUR CODE HERE
chars = ['A', 'ñ', '姚', '😀']
# Create a loop for every character in chars, the line ch.encode('utf-8') converts it into its UTF-8 byte representation. 
# Print both the actual bytes and how many bytes we need to store that character. 
for ch in chars:
    utf8_bytes = ch.encode('utf-8')
    print(f"'{ch}' -> bytes: {utf8_bytes}, length: {len(utf8_bytes)}")
````

### Exercise 3.3: String Length vs Byte Length

The slides note that len() returns code points, not bytes.
Compare string length vs byte length for mixed text.

````python
# YOUR CODE HERE
mixed_text = "Señor 你好 😀"

# The len(mixed_text) gives us the exact number of characters it shows no matter how many bytes it uses.
char_length = len(mixed_text)

# The len(mixed_text.encode("utf-8")) shows us the number of bytes we need to store it in UTF-8, this could be more than it shows because different characters can take an unequal amount of bytes.
byte_length = len(mixed_text.encode("utf-8"))

print("Text:", mixed_text)
print("Character length:", char_length)
print("Byte length (UTF-8):", byte_length)
````
